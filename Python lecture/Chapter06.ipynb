```python
while True:
    try: #예외처리
        age = int(input("나이입력: "))
    except: #예외 발생시
        print("입력이 정확하지 않습니다.")
    else: #예외가 발생하지 않을 시
        if age<=18:
            print("미성년자 출입금지")
        else:
            print("환영합니다")
        if age==0:
            break
    finally: #예외가 생기든 말든, break가 있든말든
        print("항상 실행")
```

```python
class NotThree(Exception):
    def __init__(self,e):
        super().__init__(e)
def three_mul():
    try:
        x = int(input("3의 배수를 입력하시오: "))
        if x%3!=0:
            raise NotThree("3의배수가 아니자나!")
        else:
            print("3의 배수가 맞다!")
    except NotThree as e:
        print(e)
n=0
while n<=5:
    three_mul()
    n+=1
```

```python
class BalanceInsufficientException(Exception):
    def __init__(self,e):
        super().__init__(e)

class Account:
    def __init__(self):
        self.balance = 0
    def deposit(self,money):
        self.balance += money
    def getBalance(self):
        return self.balance
    def withdraw(self,money):
        if self.balance<money:
            raise BalanceInsufficientException(f'잔고부족({money-self.balance}원 모자람)')
        else:
            self.balance -= money
            print(f'잔액: {self.balance}')

account = Account()
account.deposit(10000)
print(f'예금액: {account.getBalance()}')
try:
    account.withdraw(30000)
except BalanceInsufficientException as e:
    print(e)

print(f'{"":=^30}')
try:
    account.withdraw(5000)
except BalanceInsufficientException as e:
    print(e)
```
