## 계산기 만들기

```python
from tkinter import*
class Calculator:  #계산기 클래스
    def __init__(self): 
        self.numList = [] 
        self.result = 0 
    def setFomula(self,s): 
        if type(s)!=int: 
            if len(self.numList)==0:
                self.numList.append(str(self.result))
            elif str(self.numList).find("+"or"-"or"*"or"/"or"%")!=-1: 
                self.res()
                self.numList.append(str(self.result))
            self.numList.append(s)
            lbl['text'] = ''.join(self.numList)
        else: 
            self.numList.append(str(s))
            lbl['text'] = ''.join(self.numList)
    def res(self):
        try:
            s = ''.join(self.numList)
            self.result = round(eval(s),4)
        except Exception as e:
            lbl['text'] = "Uncompleted\nFomula"
        else:
            lbl['text'] = self.result
            self.numList=[]
    def clr(self): 
        lbl['text'] = "CLEAR" 
        self.numList=[]
        self.result=0

win=Tk()
win.title("계산기") 
win.geometry("172x220+0+0") #창크기 설정
win.resizable(False,False) #응 조절못해
win.config(bg="white")

cal = Calculator() #계산기 클래스 객체 생성

# 출력화면, 숫자, 연산기호 버튼객체 생성
lbl = Label(win, text="CALCULATOR", font=("Bank Gothic Medium BT",15), bg="white", bd=2, height=2)
bt0 = Button(win, text="0", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(0))
bt1 = Button(win, text="1", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(1))
bt2 = Button(win, text="2", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(2))
bt3 = Button(win, text="3", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(3))
bt4 = Button(win, text="4", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(4))
bt5 = Button(win, text="5", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(5))
bt6 = Button(win, text="6", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(6))
bt7 = Button(win, text="7", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(7))
bt8 = Button(win, text="8", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(8))
bt9 = Button(win, text="9", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula(9))
bt_sum = Button(win, text="+", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula("+"))
bt_sub = Button(win, text="-", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula("-"))
bt_mul = Button(win, text="x", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula("*"))
bt_div = Button(win, text="/", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula("/"))
bt_rem = Button(win, text="%", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=lambda: cal.setFomula("%"))
bt_res = Button(win, text="=", font=("Bank Gothic Medium BT",12), width=3, bg="white", command=cal.res)
bt_clr = Button(win, text="clear", font=("Bank Gothic Medium BT",12), bg="white", command=cal.clr)

#각 객체의 자리를 잡아주는 과정
lbl.grid(row=0,column=0,columnspan=4, pady=10)
bt7.grid(row=1,column=0)
bt8.grid(row=1,column=1)
bt9.grid(row=1,column=2)
bt_sum.grid(row=1,column=3)
bt4.grid(row=2,column=0)
bt5.grid(row=2,column=1)
bt6.grid(row=2,column=2)
bt_sub.grid(row=2,column=3)
bt1.grid(row=3,column=0)
bt2.grid(row=3,column=1)
bt3.grid(row=3,column=2)
bt_mul.grid(row=3,column=3)
bt0.grid(row=4,column=0)
bt_div.grid(row=4,column=1)
bt_rem.grid(row=4,column=2)
bt_res.grid(row=4,column=3)
bt_clr.grid(row=5,column=0,columnspan=4)

win.mainloop()
```

## 연습문제 중 까다로웠던 문제
#### 피보나치 수열
입력한 정수 이하까지의 피보나치 수열 출력

피보나치 함수에 대한 매커니즘은 이해했는데, 입력한 정수 이하까지의 수열을 출력하는데에 대한 반복문의 범위 설정에 애를 먹었습니다.

딱 입력한 정수까지의 숫자를 저장하는 방법은 포기하고, 어차피 피보나치 수열이 20항을 넘어가면 숫자단위가 10만을 넘어가므로

적당한 숫자를 선정해 리스트에 피보나치수열을 저장하고, 입력한 정수 이상의 요소들을 제거하는 방향으로 해결했습니다.

```python
def fib(n):
    if n==0 or n==1:  #입력값이 0 또는 1이라면 1을 리턴
        return 1
    else:
        return fib(n-1)+fib(n-2) #n-1과 n-2를 입력으로 넣은 fib()함수를 합함

n = int(input("입력: "))
lst = []
lst.append(0)
for i in range(25): #범위설정을 너무 크게 하면 연산이 오래걸려서 리터럴로 설정
    lst.append(fib(i)) #피보나치 함수의 요소들을 오름차순으로 리스트에 추가
    for j in lst: #리스트를 한번 쫙 훑어서
        if j>n: #입력한 숫자보다 큰 숫자들이 저장되었다면
            lst.remove(j) #싹다 제거
print(lst)
```
