```python
#Q1
class Calculator:
    def __init__(self):
        self.value = 0
    def add(self,val):
        self.value += val
        
class UpgradeCalculator(Calculator):
    def minus(self,val):
        self.value -= val
        
cal = UpgradeCalculator()
cal.add(10)
cal.minus(7)

print(cal.value)
```

```python
#2
class MaxLimitCalculator(Calculator):
    def add(self,val):
        self.value += val
        if self.value >= 100:
            self.value = 100

cal = MaxLimitCalculator()
cal.add(50)
cal.add(60)

print(cal.value)
```

```python
#Q5
def Fibonacci(n):
    if n==1 or n==0:
        return 1
    else:
        return Fibonacci(n-1)+Fibonacci(n-2)
n = int(input("입력: "))
lst = []
lst.append(0)
for i in range(30):
    lst.append(Fibonacci(i))
    for j in lst:
        if j>n :
            lst.remove(j)
            break
print(lst)
```

```python
#Q7 한줄 구구단
def mul(n):
    lst = []
    for i in range(1,10):
        lst.append(str(n*i))
    return lst

n = int(input("구구단을 출력할 숫자를 입력하세요(2~9): "))
if not 2<=n<=9:
    print("잘못된 입력입니다.")
else:
    lst = mul(n)
    print(' '.join(lst))
    
lst = [str(i*n) for i in range(1,10)]
print(' '.join(lst))
```

```python
#Q10 사칙연산 계산기
class Calculator:
    def __init__(self,lst):
        self.lst = list(lst)
        self.s=0
    def csum(self):
        return sum(self.lst)
    def cavg(self):
        self.s = self.csum()
        return self.s/len(self.lst)
cal = Calculator([1,2,3,4,5])
print(cal.csum())
print(cal.cavg())
```

```python
#Q12 DashInsert 함수
def dashInsert(s):
    lst = list(map(int,list(s)))
    i=0
    while i<len(lst)-1:
        if type(lst[i])!=int:
            i+=1
            continue
        if lst[i]%2==0 and lst[i+1]%2==0:
            lst.insert(i+1,'*')
        elif lst[i]%2==1 and lst[i+1]%2==1:
            lst.insert(i+1, '-')
        else:
            i+=1
            continue
        i += 1
    lst = list(map(str,lst))
    lst = ''.join(lst)
    return lst
s = input("입력: ")
print(dashInsert(s))
```

```python
#Q12 DashInsert 방법 2
data = "4546793"
def dashInsert2(s):
    res = ""
    for i in range(len(s)-1):
        res += s[i]
        if int(s[i])%2==0 and int(s[i+1])%2==0:
            res += '*'
        elif int(s[i])%2==1 and int(s[i+1])%2==1:
            res += '-'
    res = res+s[len(s)-1]
    return res
print(dashInsert2(data))
```

```python
#Q13 문자열 압축하기
s = input("문자열 입력: ")
lst = []
while len(s)!=0:
    cnt=0
    if len(s)==1:
        lst.append(s)
        lst.append("1")
        s = s.lstrip(s)
        break
    for i in range(len(s)):
        if s[0]==s[i]:
            cnt+=1
        else:
            break
    lst.append(s[0])
    lst.append(str(cnt))
    s = s.lstrip(s[0])
print(''.join(lst))
```

```python
#Q13 다른 방법
def zip_str(s):
    tmp=str()
    cnt=0
    res=str()
    for i in s:
        if i!=tmp:
            if cnt!=0:
                res = res+str(cnt)
            res+=i
            tmp=i
            cnt=1
        else:
            cnt+=1
    if cnt!=0:
        res = res+str(cnt)
    return res

print(zip_str("aaabbcccccca"))
```

```python
#Q14 Duplicate numbers
def dup(n):
    for i in range(10):
        if n.find(str(i))==-1:
            return False
    s = set(n)
    if len(s)!=len(n):
        return False
    return True
n = input("숫자입력: ")
print(dup(n))
n = input("숫자입력: ")
print(dup(n))
n = input("숫자입력: ")
print(dup(n))
```
