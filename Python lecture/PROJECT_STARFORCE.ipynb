### 스타포스

```python
from tkinter import*
import random

win = Tk()
win.title("STAR FORCE")
win.config(bg="SlateGray")
win.resizable(False,False)
fnt="malgunGothic"
bgColor="SlateGray"
fgColor="Gainsboro"

STAR={0:(95,0,100),1:(90,0,110),2:(85,0,120),3:(80,0,130),4:(75,0,140),
      5:(70,0,280),6:(65,0,290),7:(60,0,300),8:(55,0,310),9:(50,0,320),
      10:(45,0,640),11:(40,0,650),12:(35,1,660),13:(30,1,670),14:(30,1,680),
      15:(30,2,1360),16:(30,2,1370),17:(30,2,1380),18:(30,3,1390),19:(30,3,1400),
      20:(30,7,2800),21:(30,7,2900),22:(3,19,3000),23:(2,29,3500),24:(1,40,4000)}

class StarForce:
    def __init__(self):
        self.star=0
        self.cnt_fail=0
        self.stacked_meso=0
    def reinforce(self):
        rd=random.randint(1,100)
        self.stacked_meso+=STAR.get(self.star)[2]
        self.startext=str()
        if self.cnt_fail==2:
            lbl_res['text']='SUCCESS!'
            self.star+=1
            self.cnt_fail=0
        else:
            if rd<=STAR.get(self.star)[1]:
                lbl_res['text']='DESTROYED'
                lbl_final_value['text']=f"{self.star}성에서 파괴"
                self.star=12
                self.cnt_fail=0
                bt_rein['state']="disabled"
            elif rd<=STAR.get(self.star)[0]:
                self.star+=1
                self.cnt_fail=0
                lbl_res['text']='SUCCESS!'
            else:
                lbl_res['text']='FAILED..'
                if 11<=self.star<15 or 16<=self.star<20 or 21<=self.star<25:
                    self.star-=1
                    self.cnt_fail+=1
        if self.cnt_fail==2:
            lbl_suc_prob['text']="100%"
            lbl_fail_prob['text']="0%"
            lbl_dst_prob['text']="0%"
        else:
            lbl_suc_prob['text']=f'{STAR.get(self.star)[0]}%'
            lbl_fail_prob['text']=f'{100-STAR.get(self.star)[0]-STAR.get(self.star)[1]}%'
            lbl_dst_prob['text']=f'{STAR.get(self.star)[1]}%'
        
        for j in range(self.star):
            self.startext+="★"
        for j in range(self.star,25):
            self.startext+="☆"
        lbl_star['text']= f"{self.startext[:5]} {self.startext[5:10]} {self.startext[10:15]} {self.startext[15:20]} {self.startext[20:25]}"
        lbl_title['text']=f'{self.star}성 -> {self.star+1}성'
        lbl_meso_demd['text']=f'{STAR.get(self.star)[2]}메소'
        lbl_totalmeso_meso['text']=f'{self.stacked_meso}메소'
    def complete(self):
        if bt_rein['state']=="disabled":
            bt_rein['state']="normal"
        else:
            lbl_totalmeso_meso['text']=str(self.stacked_meso)+"메소"
            lbl_final_value['text']=str(self.star)+"성"
            self.stacked_meso=0
            self.star=0
            self.cnt_fail=0

def enterPress(event):
    if bt_rein['state']!="disabled":
        sf.reinforce()
def endPress(event):
    sf.complete()
sf=StarForce()
lbl_star=Label(win, width=48, text='☆☆☆☆☆ ☆☆☆☆☆ ☆☆☆☆☆ ☆☆☆☆☆ ☆☆☆☆☆', font=(fnt,12), bg=bgColor, fg="Gold")
lbl_title=Label(win, width=48, text='0성 -> 1성', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_suc_txt=Label(win, width=12, text='성공확률:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_suc_prob=Label(win, width=36, text='95%', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_fail_txt=Label(win, width=12, text='실패확률:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_fail_prob=Label(win, width=36, text='5%', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_dst_txt=Label(win, width=12, text='파괴확률:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_dst_prob=Label(win, width=36, text='0%', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_meso_txt=Label(win, width=12, text='메소:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_meso_demd=Label(win, width=36, text='0메소', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_res=Label(win, width=24, text="결과", font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_totalmeso_txt=Label(win, width=12, text='누적메소:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_totalmeso_meso=Label(win, width=36, text='0메소', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_final_txt=Label(win, width=12, text='최종결과:', font=(fnt,10), bg=bgColor, fg=fgColor)
lbl_final_value=Label(win, width=36, text='0성', font=(fnt,10), bg=bgColor, fg=fgColor)
bt_rein=Button(win, command=sf.reinforce, width=12, text="강화", font=(fnt,10), bg="DarkOliveGreen", fg=fgColor, activebackground="DarkOliveGreen", activeforeground=fgColor, disabledforeground="Crimson")
bt_comp=Button(win, command=sf.complete, width=12, text="완료", font=(fnt,10), bg="DarkOliveGreen", fg=fgColor, activebackground="DarkOliveGreen", activeforeground=fgColor)

win.bind("<Return>",enterPress)
win.bind("<End>",endPress)

lbl_star.grid(row=0,column=0,columnspan=4)
lbl_title.grid(row=1,column=0,columnspan=4)
lbl_suc_txt.grid(row=2,column=0)
lbl_suc_prob.grid(row=2, column=1,columnspan=3)
lbl_fail_txt.grid(row=3, column=0)
lbl_fail_prob.grid(row=3, column=1,columnspan=3)
lbl_dst_txt.grid(row=4, column=0)
lbl_dst_prob.grid(row=4, column=1,columnspan=3)
lbl_meso_txt.grid(row=5, column=0)
lbl_meso_demd.grid(row=5, column=1,columnspan=3)
lbl_res.grid(row=6, column=0, columnspan=2)
bt_rein.grid(row=6, column=2)
bt_comp.grid(row=6, column=3)
lbl_totalmeso_txt.grid(row=7, column=0)
lbl_totalmeso_meso.grid(row=7, column=1,columnspan=3)
lbl_final_txt.grid(row=8,column=0)
lbl_final_value.grid(row=8,column=1,columnspan=3)

win.mainloop()
```
