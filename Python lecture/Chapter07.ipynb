### abs(n)
인수의 절대값을 리턴

```python
def myabs(n):
    return -n if n<0 else n
print(myabs(-2))
print(myabs(2))

print(f'{"":=^10}')
print(abs(-3))
print(abs(3))
```

### all(iterable type) 
반복 가능한 자료형을 인수로 받으며 해당 인수가 모두 참이면 True, 하나라도 거짓이면 False리턴

```python
print(all([1,2,3]))
print(all([1,2,0])) #0은 거짓을 의미
print(all(['a','b','c']))
print(all(['a','b',''])) #빈 문자열은 거짓을 의미
```

### any(iterable type)
반복 가능한 자료형을 인수로 받으며 해당 인수에서 단 하나라도 참이 존재하면 True, 모두 거짓이면 False리턴

```python
print(any(['a','b','']))
print(any([1,0,3]))
print(any([0,0,0]))
print(any(['','','']))
```

```python
def change(ch):
    asc = ord(ch)
    if 64<asc<(64+27): #a~z범위 안일 경우
        asc += 32
    elif 96<asc<(97+27):
        asc -= 32
    else: #a~Z 범위 밖일 경우
        return '잘못입력'
    return chr(asc)
print(change('a'))
print(change('Z'))
print(change('1'))
```

#### 파이썬은 따로 반환타입을 지정하지 않으므로, 하나의 함수에서 여러 타입을 반환할 수 있음

```python
def return_check(n):
    if n==1:
        return 1
    elif n==2:
        return '!'
    else:
        return [1,2]
    
print(return_check(1))
print(return_check(2))
print(return_check(3))
```

### enumerate(gradational type)
순서가 있는 자료형을 입력으로 받아 인덱스 값을 포함하는 enumerate 객체로 반환

### eval(String)

### filter(func, iterable type)
함수에 반복가능한 자료형이 입력으로 들어가 함수의 연산결과값을 반환

```python
def positive(l):
    list = []
    for i in l:
        if i>0:
            list.append(i)
    return list
print(positive([1,-1,0,3,-4,5]))

print(list(filter(lambda x : x>0, [1,-1,0,3,-4,5]))) #함수 자리에 람다식을 넣엇따
```

```python
def maxx(a,b):
    if not isinstance(a,int) or not isinstance(b,int):
        print('Wrong input')
        return None
    else:
        return a if a>=b else b

a,b = list(map(int,input("두 수 입력: ").split()))
print(maxx(a,b))
print(maxx(1,'A'))
```

### sum(list or tuple)
리스트/튜플의 모든 요소의 합을 반환

```python
def mySum(lst):
    sum = 0
    for i in lst:
        sum += i
    return sum
math = list(map(int,input("숫자 입력: ").split()))
math_sum = mySum(math)
print(f'합계 : {math_sum}\n평균 : {math_sum/len(math)}')

print()
math = list(map(int,input("입력: ").split()))
math_sum = sum(math)
print(f'합계 : {math_sum}\n평균 : {math_sum/len(math)}')
```

### type(object)
입력값의 자료형을 반환

```python
def flat(n):
    output = []
    for item in n:
        if type(item)==list:
            output+=item
            print(output)
        else:
            output.append(item)
    return output

s = [[1,2,3],[4,5,6],[7,8,9]]
print(flat(s))
```

```python
def flat(n):
    output = []
    for item in n:
        if type(item)==list:
            output+= flat(item)
        else:
            output.append(item)
    return output
s = [[1,2,3,[4]],[4,5,6],[7,8,9]]
print(flat(s))
```
