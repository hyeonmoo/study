```python
def facto(n):
    if not isinstance(n, int) or n<0:
        return None
    else:
        if n==1:
            return 1
        else:
            return n*facto(n-1)
print(facto("a"))
print(facto(5))
```

```python
class Person:
    def __init__(self,name,age):   #생성자를 통해 객체속성 추가
        self.hello = '안녕하세요'
        self.name = name
        self.age = age
    def setAdr(self,address):      #함수를 통해 객체속성 추가
        self.address = address
        
    def greeting(self):
        print(f'{self.hello} 저는 {self.name}입니다.')
hong = Person("홍길동",20)
hong.greeting()

print('이름:',hong.name)
print('나이:',hong.age)

hong.setAdr("청주시")
print('주소:',hong.address)

hong.id = 'hhong'                   #인스턴스 생성 후 객체속성 추가(해당 객체에서만 사용가능)
print('아이디:',hong.id)

dong = Person('동그라미',20)        #hong과 다른 객체이므로 id속성에 접근할 수 없음
print(dong.id)
```

```python
class Cal:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def add(self):
        return self.x+self.y
    def sub(self):
        return self.x-self.y
    def mul(self):
        return self.x*self.y
    def div(self):
        return (int)(self.x/self.y)

cal = Cal(2,3)
print(f'{cal.x} + {cal.y} = {cal.add()}')
print(f'{cal.x} - {cal.y} = {cal.sub()}')
print(f'{cal.x} x {cal.y} = {cal.mul()}')
print(f'{cal.x} / {cal.y} = {cal.div()}')
```

```python
class Cal:
    def setData(self):
        data = input('두 수를 입력하시오 : ').split()
        self.x = int(data[0])
        self.y = int(data[1])
    def getData(self):
        return self.x, self.y
    def add(self):
        return self.x+self.y
    def sub(self):
        return self.x-self.y
    def mul(self):
        return self.x*self.y
    def div(self):
        return (int)(self.x/self.y)
    
cal = Cal()
cal.setData()
x, y = cal.getData()
print(f'{x} + {y} = {cal.add()}')
print(f'{x} - {y} = {cal.sub()}')
print(f'{x} x {y} = {cal.mul()}')
print(f'{x} / {y} = {cal.div()}')
```

```python
class Person:
    def greeting(self):
        print("안녕하세용")
class Student(Person):
    def study(self):
        print("공부하기")
        
su = Student()
su.greeting() #기반클래스의 메소드 사용 가능
su.study()

print("{0:=^10}".format(""))

hong = Person()
hong.greeting()
#hong.study() --파생클래스에서만 정의되었으므로 기반클래스 객체에선 사용불가

print(issubclass(Student,Person))
```

```python
class Car:
    def __init__(self,model,color):
        self.model = model
        self.color = color
    def info(self): #기반클래스 객체에서 메소드 호출 시 해당 메소드 출력
        print(f'Model : {self.model}\nColor : {self.color}')

class CarDrive(Car):
    def info(self): #기반클래스의 메소드를 오버라이딩 -> 파생클래스 객체에서 메소드 호출 시 재정의된 메소드 출력
        print(f'내차는 {self.model}이고, {self.color}색이다')
    def changeSpeed(self,v):
        self.speed = v
        print(f'Changing speed to : {self.speed}km/h')

jeep = CarDrive('jeep', '노란') #파생 클래스 객체 생성
jeep.info() #재정의된 메소드 출력
jeep.changeSpeed(100)

print(f'{"":=^30}')
bmw = Car('bmw','red') #기반 클래스 객체 생성
bmw.info() #기반 클래스에 정의된 메소드 출력
```
