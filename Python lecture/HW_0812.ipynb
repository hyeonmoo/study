```python
from tkinter import*
import random as rd
win = Tk()
win.title("다이아몬드")
fonttype = "Bank Gothic Light BT"
bgcolor = "AliceBlue"

win.config(bg=bgcolor)
canvas = Canvas(win, bg=bgcolor, width=600, height=600, relief="solid", bd=2)
colors = ['salmon','red','pink','deeppink','coral','orangered','gold','darkkhaki','lavender','blueviolet','lightgreen','green','olive','lightseagreen','paleturquoise','steelblue','lightblue','skyblue','darkblue','bisque','saddlebrown','beige','aliceblue','white','silver','dimgray','darkslategray','black']

def diamond(event):
    x,y=event.x,event.y
    fillcolor = rd.choice(colors)
    outlinecolor = rd.choice(colors)
    canvas.create_polygon(x-15,y-30,x-30,y-10,x-10,y-10, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x-15,y-30,x-10,y-10,x,y-30, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x,y-30,x-10,y-10,x+10,y-10, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x,y-30,x+10,y-10,x+15,y-30, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x+15,y-30,x+10,y-10,x+30,y-10, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x-30,y-10,x-10,y-10,x,y+25, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x-10,y-10,x+10,y-10,x,y+25, width=2,fill=fillcolor,outline=outlinecolor)
    canvas.create_polygon(x+10,y-10,x+30,y-10,x,y+25, width=2,fill=fillcolor,outline=outlinecolor)
    bt1.config(bg=fillcolor)

lbl1 = Label(win, text="Click or Drag Canvas!", font=(fonttype,24), bg=bgcolor)
bt1 = Button(win, text="Clear", font=(fonttype,15,'bold'), command=lambda:canvas.delete("all"))

canvas.bind("<Button-1>",diamond)
canvas.bind("<B1-Motion>",diamond)

lbl1.pack(pady=5)
canvas.pack()
bt1.pack(pady=5)

win.mainloop()
```
