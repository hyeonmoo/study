```python
def tsnNumber(a):
    clap = 0
    for i in range(1,a+1):
        stri = str(i)
        for j in stri:
            intj = int(j)
            if intj!=0 and intj%3==0:
                clap+=1
    return clap
print(tsnNumber(70))
```

```python
def tsn(a):
    tsnList = [i for i in range(1,a+1)]
    cnt=0
    clap=0
    for j in tsnList:
        cnt = str(j).count('3') + str(j).count('6') + str(j).count('9')
        if cnt==1:
            clap+=1
            tsnList[tsnList.index(j)] = "짝"
        elif cnt==2:
            clap+=2
            tsnList[tsnList.index(j)] = "짝짝"
        elif cnt==3:
            clap+=3
            tsnList[tsnList.index(j)] = "짝짝짝"
    return tsnList,clap

num = int(input("숫자 입력: "))
tsnList=tsn(num)[0]
tsnNum=tsn(num)[1]
for i in tsnList:
        print(i,end=" ")
print()
print("박수 친 횟수: %d"%tsnNum)
```

```python
def add_mul(choice, *args):
    print(args)
    if choice == "add":
        res = 0
        for i in args:
            res+=i
    elif choice == "mul":
        res = 1
        for i in args:
            res*=i
    return res
print(add_mul("add",1,2,3,4,5))
print(add_mul("mul",2,3,4,5))
```

```python
#연습문제 2
def avg(*args): #매개변수의 개수를 유동적으로 입력받아 튜플로 관리
    sum=0
    for n in args: #매개변수가 담긴 args튜플에서 매개변수들을 하나하나 호출
        sum+=n
    return sum/len(args)

print(avg(95,80,90,75,70))
```

```python
from tkinter import*
class Action:
    def __init__(self,win,img1):
        self.win = win
        self.img1 = img1
    def action1(self, lbl, img):
        lbl['image'] = img
        lbl.image = img


win = Tk()
img1 = PhotoImage(file="scissor.png", master = win)
img2 = PhotoImage(file="rock.png", master = win)
img3 = PhotoImage(file="paper.png", master = win)

ch1 = Action(win, img1)
lblRSP = Label(win, text="Rock Scissor Papers",font=("Magneto",15),width=20, bg="black",fg="white")
lblcom = Label(win, text="Computer",font=("Magneto",10), bg="white")
lblplayer = Label(win, text="Player",font=("Magneto",10), bg="white")

btn1 = Button(win, width=8, text="가위",bg="white", command=lambda:ch1.action1(lblplayer,img1))
btn2 = Button(win, width=8, text="바위",bg="white", command=lambda:ch1.action1(lblplayer,img2))
btn3 = Button(win, width=8, text="보",bg="white", command=lambda:ch1.action1(lblplayer,img3))

lblRSP.grid(row=0, column=0, columnspan=4)
lblcom.grid(row=1, column=0, columnspan=2)
lblplayer.grid(row=1, column=1, columnspan=2)
btn1.grid(row=2, column=0)
btn2.grid(row=2, column=1)
btn3.grid(row=2, column=2)
win.mainloop()
```
