### 메모장 만들기

```python
from tkinter import*

def save_clk(file):
    try:
        s=txt.get(1.0,END)
        f=open(file,'w',encoding="utf-8")
        f.write(s)
        f.close()
    except Exception as e:
        lbl_state.config(text="Saving Failed")
    else:
        lbl_state.config(text="Saving Complete")
    finally:
        txt.delete(1.0,END)
        ent.delete(0,END)

def get_clk(file):
    try:
        txt.delete(1.0,END)
        f=open(file,"r",encoding="utf-8")
        m = f.read()
        f.close()
    except Exception as e:
        lbl_state.config(text="File Loading Failed")
    else:
        txt.insert(1.0,m)
        lbl_state.config(text="File Loading Complete")
    finally:
        ent.delete(0,END)

def del_clk(file):
    try:
        f=open(file,"w",encoding='utf-8')
        f.write('')
        f.close()
    except Exception as e:
        lbl_state.config(text="File Loading Failed")
    else:
        txt.delete(1.0,END)
        lbl_state.config(text="Delete Successful")
    finally:
        ent.delete(0,END)

win = Tk()
win.title("NOTEBOOK")
win.geometry("405x300+0+0")
win.config(bg="mediumseagreen")

f = "문체부 궁체 흘림체"
txt = Text(win, width=50, height=10, font=(f,13,"bold"), bg="mediumseagreen", fg="white")
lbl_file = Label(win, width=7, text="파일명:", font=(f,13,"bold"), bg="mediumseagreen")
lbl_state = Label(win, text="상태표시", width=30,height=2, font=(f,13,"bold"), bg="mediumseagreen")
ent = Entry(win, font=(f,13), width=28, bg="mediumseagreen", fg="white")

save = Button(win, text="메모저장", font=(f,10,'bold'), bg="white", fg="seagreen", height=0, command=lambda:save_clk(ent.get()))
get = Button(win, text="불러오기", font=(f,10,'bold'), bg="white", fg="seagreen", height=0, command=lambda:get_clk(ent.get()))
delete = Button(win, text="내용삭제", font=(f,10,'bold'), bg="white", fg="seagreen", height=0, command=lambda:del_clk(ent.get()))

save.grid(row=1,column=2)
get.grid(row=2,column=2)
delete.grid(row=3,column=2)

txt.grid(row=0,column=0,columnspan=3)
lbl_state.grid(row=1,column=0,rowspan=2,columnspan=2)
lbl_file.grid(row=3,column=0)
ent.grid(row=3,column=1)

win.mainloop()
```
