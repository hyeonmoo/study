### pygame 기본구조
    1) pygame import
    2) pygame.init()으로 초기화
    3) 변수 선언-초기화
    4) pygame 메인루프 설정

```python
import pygame as pg
import sys
from pygame.locals import*

pg.init()

size=[400, 300]
screen = pg.display.set_mode(size)
pg.display.set_caption("game title")
sysfont = pg.font.SysFont('mdgaesung',20)
font = pg.font.Font("C:\\Users\\green\\AppData\\Local\\Microsoft\\Windows\\Fonts\\SDSamliphopangcheOutline.otf",20)
clock = pg.time.Clock()
txt=sysfont.render("",True,(173, 216, 230))
done = True

while done:
    screen.fill((240, 248, 255))
    for event in pg.event.get():
        if event.type==pg.QUIT:
            done=False
            pg.quit()
            sys.exit()
        if event.type==KEYDOWN:
            if event.key==K_UP:
                txt=sysfont.render("UP",True,(173, 216, 230))
            elif event.key==K_DOWN:
                txt=sysfont.render("DOWN",True,(173, 216, 230))
            elif event.key==K_LEFT:
                txt=sysfont.render("LEFT",True,(173, 216, 230))
            elif event.key==K_RIGHT:
                txt=sysfont.render("RIGHT",True,(173, 216, 230))
        if event.type==MOUSEBUTTONDOWN:
            if event.button==1:
                txt=font.render("마우스왼쪽버튼", True, (216, 191, 216))
            elif event.button==2:
                txt=font.render("마우스휠버튼", True, (216, 191, 216))
            elif event.button==3:
                txt=font.render("마우스오른쪽버튼", True, (216, 191, 216))
    screen.blit(txt,(150,150))
    pg.display.update()
    clock.tick(60)
```

```python
import pygame as g
from PIL import Image

SCREEN_WIDTH=400
SCREEN_HEIGHT=500
size=[SCREEN_WIDTH,SCREEN_HEIGHT]
g.init()
SCREEN=g.display.set_mode(size)
g.display.set_caption("pygame test")
clock=g.time.Clock()

teemo = Image.open("Teemo.png")
teemo_size=teemo.resize((100,100))
teemo_size.save("Teemo2.png")

player=g.image.load("Teemo2.png")
player_Rect=player.get_rect()
playing=True
x,y=100,100
dx,dy=0,0
while playing:
    SCREEN.fill((255,255,255))
    for event in g.event.get():
        if event.type==g.QUIT:
            playing=False
            g.quit()
        if event.type==g.KEYDOWN:
            if event.key==g.K_UP:
                dy=-5
            elif event.key==g.K_DOWN:
                dy=5
            elif event.key==g.K_LEFT:
                dx=-5
            elif event.key==g.K_RIGHT:
                dx=5
        if event.type==g.KEYUP:
            if event.key==g.K_UP:
                dy=0
            elif event.key==g.K_DOWN:
                dy=0
            elif event.key==g.K_LEFT:
                dx=0
            elif event.key==g.K_RIGHT:
                dx=0
    x+=dx
    y+=dy
    if x<0:
        x=0
    if y<0:
        y=0
    if x>SCREEN_WIDTH-player_Rect.w:
        x=SCREEN_WIDTH-player_Rect.w
    if y>SCREEN_HEIGHT-player_Rect.h:
        y=SCREEN_HEIGHT-player_Rect.h
    
    SCREEN.blit(player,[x,y])
    g.display.update()
    clock.tick(60)
```

```python
import pygame as g
import sys
from pygame.locals import*
import random as rd

SCREEN_WIDTH=600
SCREEN_HEIGHT=600
size=[SCREEN_WIDTH,SCREEN_HEIGHT]
g.init()
SCREEN=g.display.set_mode(size)
g.display.set_caption("방향키조작")
CLOCK=g.time.Clock()
font = g.font.Font("C:\\Users\\green\\AppData\\Local\\Microsoft\\Windows\\Fonts\\SDSamliphopangcheOutline.otf",30)

title=font.render("SCREEN",True,(0,0,0))
title_rec=title.get_rect()
title_rec.centerx,title_rec.centery=300,150

up_rect=Rect(250,250,100,60)
right_rect=Rect(380,300,100,60)
left_rect=Rect(120,300,100,60)
down_rect=Rect(250,350,100,60)

up_text=font.render("UP",True,(255,255,255))
up_text_rec=up_text.get_rect()
up_text_rec.centerx,up_text_rec.centery=up_rect.centerx,up_rect.centery

right_text=font.render("RIGHT",True,(255,255,255))
right_text_rec=right_text.get_rect()
right_text_rec.centerx,right_text_rec.centery=right_rect.centerx,right_rect.centery

left_text=font.render("LEFT",True,(255,255,255))
left_text_rec=left_text.get_rect()
left_text_rec.centerx,left_text_rec.centery=left_rect.centerx,left_rect.centery

down_text=font.render("DOWN",True,(255,255,255))
down_text_rec=down_text.get_rect()
down_text_rec.centerx,down_text_rec.centery=down_rect.centerx,down_rect.centery

right_Text=font.render("RIGHT",True,(255,255,255))
left_Text=font.render("LEFT",True,(255,255,255))
down_Text=font.render("BOTTOM",True,(255,255,255))

rudder=['UP','DOWN','LEFT','RIGHT']
cnt=0
GREEN=(152, 251, 152)
BLUE=(173, 216, 230)
RED=(250, 128, 114)
upColor,rightColor,leftColor,downColor=GREEN,GREEN,GREEN,GREEN
run=True
while run:
    SCREEN.fill((255,255,255))
    
    if cnt>=60:
        cnt=0
        random_rudder = rd.choice(rudder)
        title = font.render(random_rudder,True,(0,0,0))
    
    for event in g.event.get():
        if event.type==g.QUIT:
            run=False
            g.quit()
            sys.exit()
        if event.type==KEYDOWN:
            if event.key==K_UP:
                if random_rudder=="UP":
                    upColor=BLUE
                else:
                    upColor=RED
            if event.key==K_RIGHT:
                if random_rudder=="RIGHT":
                    rightColor=BLUE
                else:
                    rightColor=RED
            if event.key==K_LEFT:
                if random_rudder=="LEFT":
                    leftColor=BLUE
                else:
                    leftColor=RED
            if event.key==K_DOWN:
                if random_rudder=="DOWN":
                    downColor=BLUE
                else:
                    downColor=RED
        if event.type==KEYUP:
            if event.key==K_UP:
                upColor=GREEN
            if event.key==K_RIGHT:
                rightColor=GREEN
            if event.key==K_LEFT:
                leftColor=GREEN
            if event.key==K_DOWN:
                downColor=GREEN
    
    g.draw.rect(SCREEN,upColor,up_rect)
    g.draw.rect(SCREEN,rightColor,right_rect)
    g.draw.rect(SCREEN,leftColor,left_rect)
    g.draw.rect(SCREEN,downColor,down_rect)

    SCREEN.blit(up_text,up_text_rec)
    SCREEN.blit(right_text,right_text_rec)
    SCREEN.blit(left_text,left_text_rec)
    SCREEN.blit(down_text,down_text_rec)
    SCREEN.blit(title,title_rec)

    g.display.update()
    cnt+=1
    CLOCK.tick(60)
```
