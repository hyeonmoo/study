## 계산기 만들기

```python
from tkinter import*

class Calculator:
    def __init__(self):
        self.cal_state=str() #수식을 담을 문자열 생성
        self.result=0 #결과값을 담을 변수 생성
    def set_fomula(self,s):
        if type(s)==str: #연산기호가 입력됐을 때
            if self.cal_state.find("+")==-1 and self.cal_state.find("-")==-1 and self.cal_state.find("*")==-1 and self.cal_state.find("/")==-1 and self.cal_state.find("%")==-1:
                self.cal_state+=s #기입력된 연산기호가 없을 시 실행
                state['text']=self.cal_state
            elif s==".": #소수점이 입력되었을때(연산기호와 혼동방지)
                self.cal_state+=s
                state['text']=self.cal_state
            else: #기입력된 연산기호가 있을 시
                self.res() #결과처리 후 연산기호를 붙임
                self.cal_state+=s
                state['text']=self.cal_state
        elif self.cal_state==str(self.result): #연산완료후 새로운 숫자를 입력할 시 문자열초기화
            self.cal_state=str(s)
            state['text']=self.cal_state
        else: #일반적인 숫자 입력시
            self.cal_state+=str(s) #eval()로 연산하기 위해 문자열 형태로 추가
            state['text']=self.cal_state 
    def res(self):
        try:
            self.result=round(eval(self.cal_state),6) #문자열 내용을 연산해 결과값으로 저장
        except Exception as e: #완전한 수식이 아닌데 결과값을 얻으려할때
            state['text']="Uncompleted\nFomula"
        else:
            state['text']=self.result #결과값 출력
            self.cal_state=str(self.result) #수식문자열에 결과값을 저장해 이후 연산기호 입력 시 추가연산 진행
    def delete(self):
        self.cal_state=self.cal_state[:-1]
        state['text']=self.cal_state
    def clr(self):
        self.cal_state=str()
        self.result=0
        state['text']="CLEAR"

win=Tk()
win.title("계산기")
win.config(bg="HoneyDew")
win.resizable(False,False)
calcfont="malgungothic"

cal = Calculator()

def onKeyPress(event):
    cal_num="0123456789"
    cal_op="+-*/%."
    i=event.char
    if cal_num.find(i)!=-1:
        cal.set_fomula(int(i))
    elif cal_op.find(i)!=-1:
        cal.set_fomula(i)
def enterPress(event):
    cal.res()
def delPress(event):
    cal.delete()
def escPress(event):
    cal.clr()

state = Label(win, text="CALCULATOR", font=(calcfont,15,'bold'), fg="DodgerBlue", width=15, height=5, bg="HoneyDew")
bt0 = Button(win, text="0", command=lambda:cal.set_fomula(0), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt1 = Button(win, text="1", command=lambda:cal.set_fomula(1), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt2 = Button(win, text="2", command=lambda:cal.set_fomula(2), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt3 = Button(win, text="3", command=lambda:cal.set_fomula(3), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt4 = Button(win, text="4", command=lambda:cal.set_fomula(4), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt5 = Button(win, text="5", command=lambda:cal.set_fomula(5), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt6 = Button(win, text="6", command=lambda:cal.set_fomula(6), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt7 = Button(win, text="7", command=lambda:cal.set_fomula(7), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt8 = Button(win, text="8", command=lambda:cal.set_fomula(8), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
bt9 = Button(win, text="9", command=lambda:cal.set_fomula(9), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btd = Button(win, text=".", command=lambda:cal.set_fomula("."), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")

btP = Button(win, text="+", command=lambda:cal.set_fomula("+"), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btM = Button(win, text="-", command=lambda:cal.set_fomula("-"), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btX = Button(win, text="X", command=lambda:cal.set_fomula("*"), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btD = Button(win, text="/", command=lambda:cal.set_fomula("/"), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btR = Button(win, text="%", command=lambda:cal.set_fomula("%"), font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btE = Button(win, text="=", command=cal.res, font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btDel = Button(win, text="DEL", command=cal.delete, font=(calcfont,12,'bold'), width=4, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")
btC = Button(win, text="CLEAR", command=cal.clr, font=(calcfont,12,'bold'), width=9, height=1, fg="DodgerBlue", bg="HoneyDew", bd=0, relief="flat", overrelief="flat", activebackground="LightBlue",activeforeground="HoneyDew")

win.bind('<Key>',onKeyPress)
win.bind('<Return>',enterPress)
win.bind('<BackSpace>',delPress)
win.bind('<Delete>',delPress)
win.bind('<Escape>',escPress)

state.grid(row=0,column=0,columnspan=4,pady=5)
btP.grid(row=1,column=0, padx=0,pady=0)
btM.grid(row=1,column=1, padx=0,pady=0)
btX.grid(row=1,column=2, padx=0,pady=0)
btDel.grid(row=1,column=3, padx=0,pady=0)
bt7.grid(row=2,column=0, padx=0,pady=0)
bt8.grid(row=2,column=1, padx=0,pady=0)
bt9.grid(row=2,column=2, padx=0,pady=0)
btD.grid(row=2,column=3, padx=0,pady=0)
bt4.grid(row=3,column=0, padx=0,pady=0)
bt5.grid(row=3,column=1, padx=0,pady=0)
bt6.grid(row=3,column=2, padx=0,pady=0)
btR.grid(row=3,column=3, padx=0,pady=0)
bt1.grid(row=4,column=0, padx=0,pady=0)
bt2.grid(row=4,column=1, padx=0,pady=0)
bt3.grid(row=4,column=2, padx=0,pady=0)
btE.grid(row=4,column=3, padx=0,pady=0)
bt0.grid(row=5,column=0, padx=0,pady=0)
btd.grid(row=5,column=1, padx=0,pady=0)
btC.grid(row=5,column=2,columnspan=2, padx=0,pady=0)

win.mainloop()
```
