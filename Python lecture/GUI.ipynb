### 예제) 가위바위보 게임

```python
from tkinter import*
import random as rd
def action(img,player):
    player_lbl['image']=img
    RSP = ("가위","바위","보")
    com = rd.choice(RSP)
    if com=="가위":
        com_lbl['image']=scissor
    elif com=="바위":
        com_lbl['image']=rock
    else:
        com_lbl['image']=paper
    
    if player==com:
        res_lbl['text']="DRAW"
    else:
        if com=="가위":
            if player=="바위": res_lbl['text']="WIN!"
            else: res_lbl['text']="LOSE.."
        elif com=="바위":
            if player=="보": res_lbl['text']="WIN!"
            else: res_lbl['text']="LOSE.."
        else:
            if player=="가위": res_lbl['text']="WIN!"
            else: res_lbl['text']="LOSE.."
    
win=Tk()
win.title("가위바위보 게임")
win.geometry("900x300+0+0")

com = PhotoImage(file="com.png", master=win)
player = PhotoImage(file="player.png",master=win)
scissor = PhotoImage(file="scissor.PNG", master=win)
rock = PhotoImage(file="rock.PNG", master=win)
paper = PhotoImage(file="paper.PNG", master=win)

lblc = Label(win, text="COMPUTER", font=("Sandoll 삼립호빵체 Outline",15),width=25)
lblp = Label(win, text="PLAYER", font=("Sandoll 삼립호빵체 Outline",15),width=25)
com_lbl = Label(win, image=com, bd=3, width=150, height=150, relief="solid")
player_lbl = Label(win, image=player, bd=3, width=150, height=150, relief="solid")
res_lbl = Label(win, text="RESULT", font=("Sandoll 삼립호빵체 Outline",15),width=25, bg="yellow")

scis_btn = Button(win, text="가위", bg = "skyblue", font=("Sandoll 삼립호빵체 Basic",12),width=20,command=lambda: action(scissor,"가위"))
rock_btn = Button(win, text="바위", bg = "skyblue", font=("Sandoll 삼립호빵체 Basic",12),width=20,command=lambda: action(rock,"바위"))
paper_btn = Button(win, text="보", bg = "skyblue", font=("Sandoll 삼립호빵체 Basic",12),width=20,command=lambda: action(paper,"보"))

lblc.grid(row=0,column=0,padx=5,pady=5)
lblp.grid(row=0,column=2,padx=5,pady=5)
com_lbl.grid(row=1,column=0,padx=5,pady=5)
res_lbl.grid(row=1,column=1,padx=5,pady=5)
player_lbl.grid(row=1,column=2,padx=5,pady=5)
scis_btn.grid(row=2,column=0,padx=5,pady=5)
rock_btn.grid(row=2,column=1,padx=5,pady=5)
paper_btn.grid(row=2,column=2,padx=5,pady=5)

win.mainloop()
```

### 텍스트박스 예제-메모장 만들기

```python
from tkinter import*
def save_click(filename):
    try:
        s = txt.get(1.0,END)
        f=open(filename,"w",encoding='utf-8')
        f.write(s)
        f.close()
    except Exception as e:
        lbl_result.config(text="파일을 저장하는데 실패했습니다.")
    else:
        lbl_result.config(text="메모장 저장 완료")
    finally:
        txt.delete(1.0,END)
        entry.delete(0,END)

def get_click(filename):
    try:
        txt.delete(1.0, END)
        f=open(filename,"r",encoding='utf-8')
        memo = f.read()
        f.close()
    except Exception as e:
        lbl_result.config(text="파일을 불러오는데 실패했습니다.")
    else:
        txt.insert(1.0, memo)
        lbl_result.config(text="메모 불러오기 완료")
    finally: 
        entry.delete(0,END)

def delete_click(filename):
    try:
        f=open(filename,"w", encoding='utf-8')
        f.write('')
        f.close()
    except Exception as e:
        lbl_result.config(text="파일을 불러오는데 실패했습니다.")
    else:
        txt.delete(1.0, END)
        lbl_result.config(text="파일내용 삭제 완료")
    finally:
        entry.delete(0,END)
    
win = Tk()
win.title("나만의 메모장")
win.geometry("605x125+0+0")
win.config(bg="lightgray")

f = "카페24 고운밤"
txt = Text(win, font=(f,10), width=100, height=5)
lbl_filename = Label(win, text="FileName : ", font=(f,10),bg="lightgray")
entry = Entry(win, text="", font=(f,10), width=70)

save = Button(win, text="SAVE", font=(f,10), width=4, bg="LawnGreen", fg="Green", command=lambda:save_click(entry.get()))
get = Button(win, text="GET", font=(f,10), width=4, bg="Turquoise", fg="Blue", command=lambda:get_click(entry.get()))
delete = Button(win, text="DEL", font=(f,10), width=4, bg="Salmon", fg="Red", command=lambda:delete_click(entry.get()))
lbl_result = Label(win, text="", font=(f,10), width=100, bg="lightgray")

txt.grid(row=0, column=0, columnspan=5)
lbl_filename.grid(row=1, column=0)
entry.grid(row=1, column=1)
save.grid(row=1, column=2)
get.grid(row=1, column=3)
delete.grid(row=1, column=4)
lbl_result.grid(row=2, column=0, columnspan=5)

win.mainloop()
```

### 캔버스 예제- 그림판 만들기

```python
from tkinter import*
class DrawingBoard:
    def __init__(self):
        self.color="dimgray"
        self.width=5
    def setColor(self,color):
        self.color=color
    def changeWidth(self,b):
        if b:
            self.width += 2
        else:
            self.width -= 2
    def paint(self,event):
        x,y = event.x, event.y
        x2,y2,=x+self.width,y+self.width
        canvas.create_oval(x,y,x2,y2,width=self.width, outline=self.color)
    def clr(self):
        canvas.delete("all")
win = Tk()
win.config(bg="GhostWhite")
db = DrawingBoard()
f='카페24 아네모네에어'
canvas = Canvas(win, width=300, height=300, bg="white")
bt1 = Button(win, width=4, text="gray", font=(f,11,'bold'), bg="dimgray", fg='white', command=lambda:db.setColor("dimgray"))
bt2 = Button(win, width=4, text="red", font=(f,11,'bold'), bg="salmon", fg='white', command=lambda:db.setColor("salmon"))
bt3 = Button(win, width=4, text="blue", font=(f,11,'bold'), bg="lightblue", fg="white", command=lambda:db.setColor("lightblue"))
bt4 = Button(win, width=4, text="green", font=(f,11,'bold'), bg="SeaGreen", fg="white", command=lambda:db.setColor("SeaGreen"))
bt5 = Button(win, width=4, text="+", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=lambda:db.changeWidth(True))
bt6 = Button(win, width=4, text="-", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=lambda:db.changeWidth(False))
bt7 = Button(win, width=8, text="CLEAR", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=db.clr)
canvas.bind("<B1-Motion>",db.paint)
canvas.grid(row=0,column=0, columnspan=4)
bt1.grid(row=1,column=0, pady=2)
bt2.grid(row=1,column=1, pady=2)
bt3.grid(row=1,column=2, pady=2)
bt4.grid(row=1,column=3, pady=2)
bt5.grid(row=2,column=0)
bt6.grid(row=2,column=1)
bt7.grid(row=2,column=2, columnspan=2)
win.mainloop()
```

```python
from tkinter import*
class DrawingBoard:
    def __init__(self):
        self.color="dimgray"
        self.width=5
        self.drawMode="pen"
    def drawingMode(self,t):
        self.drawMode = t
    def setColor(self,color):
        if self.drawMode=="pen":
            self.color=color
        elif self.drawMode=="canvas":
            canvas['bg']=color
    def changeWidth(self,b):
        if b:
            self.width += 2
            if self.width>50:
                self.width=50
        else:
            self.width -= 2
            if self.width<4:
                self.width=4
    def paint(self,event):
        x,y = event.x, event.y
        x2,y2,=x+self.width,y+self.width
        canvas.create_oval(x,y,x2,y2,width=self.width, outline=self.color)
    def clr(self):
        if self.drawMode == "canvas":
            canvas['bg'] = "white"
        canvas.delete("all")
win = Tk()
win.title("그림판")
win.config(bg="GhostWhite")
db = DrawingBoard()
f='카페24 고운밤'
canvas = Canvas(win, width=300, height=300, bg="white")
bt1 = Button(win, width=4, text="gray", font=(f,11,'bold'), bg="dimgray", fg='white', command=lambda:db.setColor("dimgray"))
bt2 = Button(win, width=4, text="red", font=(f,11,'bold'), bg="salmon", fg='white', command=lambda:db.setColor("salmon"))
bt3 = Button(win, width=4, text="blue", font=(f,11,'bold'), bg="lightblue", fg="white", command=lambda:db.setColor("lightblue"))
bt4 = Button(win, width=4, text="green", font=(f,11,'bold'), bg="SeaGreen", fg="white", command=lambda:db.setColor("SeaGreen"))
bt5 = Button(win, width=4, text="+", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=lambda:db.changeWidth(True))
bt6 = Button(win, width=4, text="-", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=lambda:db.changeWidth(False))
bt7 = Button(win, width=8, text="CLEAR", font=(f,11,'bold'), bg="WhiteSmoke", fg="DarkSlateGray", command=db.clr)
canvas.bind("<B1-Motion>",db.paint)
radio1 = Radiobutton(win, text="Canvas",font=(f,10), bg="GhostWhite", width=5, value="canvas", command=lambda:db.drawingMode("canvas"))
radio2 = Radiobutton(win, text="Pen",font=(f,10), bg="GhostWhite", width=5, value="pen", command=lambda:db.drawingMode("pen"))

radio1.deselect()
radio2.select()

canvas.grid(row=0,column=0, columnspan=4)
radio1.grid(row=1,column=0, pady=2)
radio2.grid(row=1,column=2, pady=2)
bt1.grid(row=2,column=0, pady=2)
bt2.grid(row=2,column=1, pady=2)
bt3.grid(row=2,column=2, pady=2)
bt4.grid(row=2,column=3, pady=2)
bt5.grid(row=3,column=0)
bt6.grid(row=3,column=1)
bt7.grid(row=3,column=2, columnspan=2)
win.mainloop()
```

```python
from tkinter import*
state = "<B1-Motion>"
def paint(event):
    x,y=event.x,event.y
    canvas.create_line(x,y,x+5,y+5, width=5, fill="pink")
def change():
    global state
    if state == "<B1-Motion>":
        canvas.unbind("<B1-Motion>")
        state = "<Button-1>"
        canvas.bind(state, paint)
    else:
        canvas.unbind("<Button-1>")
        state = "<B1-Motion>"
        canvas.bind(state, paint)
win = Tk()
win.title("unbind exercise")
ft = "Bank Gothic Light BT"
canvas = Canvas(win, bg="white")
button = Button(win, text="Change mode", font=(ft,10), command=change)
canvas.bind(state,paint)
canvas.pack()
button.pack()

win.mainloop()
```
