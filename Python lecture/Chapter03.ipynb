```python
import random as rd
rsp = ['가위','바위','보']
run = 0
playerwin=0
comwin=0
print(f'{"가위바위보 게임":=^27}')
while run<5:
    if playerwin==3 or comwin==3: break
    com = rd.choice(rsp)
    player = input("가위 바위 보!>>")
    if player!='가위' and player!='바위' and player!='보':
        print("똑바로 내!")
        continue
    if com==player:
        print("Com : %s" %com)
        print(f'{"비김!":=^32}')
        continue
    else:
        if com=='가위':
            print("Com : '가위'")
            if player=='바위':
                playerwin += 1
                print(f'{"플레이어 승!":=^29}')
                run += 1
            else:
                comwin += 1
                print(f'{"컴퓨터 승!":=^30}')
                run += 1
        elif com=='바위':
            print("Com : '바위'")
            if player=='보':
                playerwin += 1
                print(f'{"플레이어 승!":=^29}')
                run += 1
            else:
                comwin += 1
                print(f'{"컴퓨터 승!":=^30}')
                run += 1
        elif com=='보':
            print("Com : '보'")
            if player=='가위':
                playerwin += 1
                print(f'{"플레이어 승!":=^29}')
                run += 1
            else:
                comwin += 1
                print(f'{"컴퓨터 승!":=^30}')
                run += 1
        
print(f'{"최종 스코어":=^29}')
print(f'{"%d : %d":^35}' %(playerwin,comwin))
if playerwin > comwin:
    print(f'{"승리!":=^32}')
else:
    print(f'{"패배ㅠ":=^31}')
```
